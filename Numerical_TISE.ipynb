```python
import numpy as np
import matplotlib.pyplot as plt

#Numerical Hamiltonian
def ham(x_pos, pot_array):
  N = len(x_pos)
  dx = x_pos[1] - x_pos[0]
  potential = np.diag(pot_array)
  kinetic = -((np.eye(N, N, k=+1) -2*np.eye(N, N, k =0) + np.eye(N, N, k=-1))/(dx**2))/2
  hamiltonian_matrix = kinetic + potential
  return hamiltonian_matrix

#Double Well Potential

N = 500 #Number of points
L = 10 #Length of the "ISQ" potential
x_pos = np.linspace(-L/2, L/2, N)


def V(v0, x_pos , x_a, N):
  v  = np.zeros(N)
  for j in range(N):
    if -x_a <= x_pos[j] <= x_a: #the range of the perturbing V0
      v[j] = v0
  return v

finitewell = V(5, x_pos , 2, N)
hamiltonian = ham(x_pos, finitewell)
eigenvalues, eigenvectors = np.linalg.eigh(hamiltonian)


for i in range(10):
    plt.plot(x_pos, eigenvalues[i] + eigenvectors[:,i], label = r"$E_{%s}$" %i)


finitewell[0]= finitewell[N-1]=100
plt.plot(x_pos, finitewell, "k--", label =r"$V_0$")
plt.legend()
plt.ylim(0,8)
plt.xlim(-10, 10)
plt.xlabel(r"$x$")
plt.ylabel(r"$V(x)$")
plt.title("Double Well Potential")



```
