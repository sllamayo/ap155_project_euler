```python
import numpy as np
import matplotlib.pyplot as plt

#Problem 3.8.a
def f(x): #Analytical Function of f(x)
  return (1.0 - np.cos(x))/x**2.0

def fpp(x): #Analytical Function of f"(x)
  return (6.0 - 4.0*x*np.sin(x) + (x**2.0 - 6.0)*np.cos(x))/x**4.0

print(fpp(0.004))

#Problem 3.8.b

def fc2(x, h): #Numerical approximation of f"(x) Using Central Difference Method
  return (f(x + h) - 2*f(x) + f(x - h))/(h**2)

hvalues = np.logspace(-1, -6, 128)

#Problem 3.8.c

def g(x): #Analytical Function with the use of trigonometric identity
  return (2*(np.sin(x/2))**2)/x**2

def gc2(x, h): #Numerical approximation of g"(x)
  return (g(x + h) - 2*g(x) + g(x - h))/(h**2)

plt.scatter (hvalues, abs(fc2(0.004, hvalues) - fpp(0.004)), s = 10, label =  r'Error: $f(x) =\frac{1-cos(x)}{x^2}$')
plt.scatter (hvalues, abs(gc2(0.004, hvalues) - fpp(0.004)), s = 10, label = r'Error: $g(x) = \frac{2 sin^2(x/2)}{x^2}$')
plt.legend()
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Δx')
plt.ylabel('Absolute Difference')
plt.title('Absolute Difference vs Δx')


```

```python


#Problem 3.25.a

def fnoise(x):
  return 2 + 5*np.sin(x) + 0.1*np.sin(30*x)

def gnoise(x):
  return  2 + 5*np.sin(x)

testx = np.linspace(0 , 2*np.pi, 128)

plt.plot(testx, fnoise(testx), label = r'f(x) =2 + 5sin(x) + 0.1sin(30x)')
plt.plot(testx, gnoise(testx), label  = r'g(x) = 2 + 5sin(x)')
plt.legend()
plt.title("Plots of Noisy Functions")

```

```python
#Problem 3.25.b
def dfnoise(x):
  return 5*np.cos(x) + 0.1*30*np.cos(30*x)

def dgnoise(x):
  return 5*np.cos(x)

def num_dfnoise(x, h):
  return ((fnoise(x + h) - fnoise(x))/(h))

plt.plot(testx, dfnoise(testx), label =  r"Analytical f'(x)")
plt.plot(testx, dgnoise(testx), label =  r"Analytical g'(x)")
plt.plot(testx, num_dfnoise(testx,0.1), label = r"Numerical f'(X) for h = 0.1")
plt.plot(testx, num_dfnoise(testx,0.01), label = r"Numerical f'(X) for h = 0.01")
plt.legend()
plt.title('Analytical vs Numerical Differentiation (Δx = h)')
```

```python
#Problem 3.25.c
def dfnoise(x):
  return 5*np.cos(x) + 0.1*30*np.cos(30*x)

def dgnoise(x):
  return 5*np.cos(x)

def num_dfnoise2(x, h):
  return ((fnoise(x + 2*h) - fnoise(x))/(2*h))

plt.plot(testx, dfnoise(testx), label = r"Analytical f'(x)")
plt.plot(testx, dgnoise(testx), label = r"Analytical g'(x)")
plt.plot(testx, num_dfnoise2(testx,0.1), label = r"Numerical g'(X) for h = 0.1")
plt.plot(testx, num_dfnoise2(testx,0.01), label = r"Numerical f'(X) for h = 0.01")
plt.legend()
plt.title('Analytical vs Numerical Differentiation (Δx = 2h)')
```
