```python
import random

#A1 Task

def random_integer_matrix(nrows, ncols):

  j = 0
  k = 0

  matrix = []
  while j < nrows:
    columns = []
    i = 0
    while i < ncols:
      k = random.randint(1,9)
      columns.append(k)
      i += 1
    matrix.append(columns)
    j += 1
  return matrix

#A2 Task

def scalar_product(list1, list2):

  sum = 0
  k = 0
  i = 0

  while i < len(list1):
    k = list1[i]*list2[i]
    sum += k
    i += 1

  return sum

#A3 Task

def transpose_matrix(orig_matrix):
  newrows = []
  newmatrix = []
  ncols = len(orig_matrix)
  nrows = len(orig_matrix[0])
  i = 0
  j = 0
  while j < nrows:
    newrows = []
    i = 0
    while i < ncols:
      newrows.append(orig_matrix[i][j])
      i += 1
    newmatrix.append(newrows)
    j += 1

  return newmatrix

#A4

```
